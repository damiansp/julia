```julia
using Javis
```

```julia
function planet(radius=15, color="green"; action=:fill, p=0)
    return function(video, object, frame)
        sethue(color)
        circle(p, radius, action)
        return p
    end
end
```

```julia
function orbit(radius, color="orchid1"; p=0)
    return function(video, object, frame)
        sethue(color)
        circle(p, radius, :stroke)
        return p
    end
end
```
