```julia
function ginger(x, y, a)
    x2 = 1. - y + a*abs(x)
    #y2 = x
    x2, x
end
```

```julia
x = [20.]
y = [9.]
for i in 1:5000
    x2, y2 = ginger(x[end], y[end], 1.74)
    push!(x, x2)
    push!(y, y2)
end
```

```julia
p1 = plot(
    sin, 
    cos, 
    0, 
    2n; 
    title="Circle",
    ratio=1,
    grid=false,
    ticks=false,
    legend=false)
p2 = plot(
    x -> x^2,
    -1,
    1;
    title="Parabola",
    gridalpha=0.4,
    gridstyle=:dot,
    legend=false)
plot(p1, p2; plot_title="Conics", plot_titlefontsize=20)
```

```julia
plot()
for n = 1:5
    plot!(x -> x^n; lw=3, label=n, color=n)
end
plot!(; legend=:topleft, legendtitle="Exponent")
```

```julia
plot()
for n = 1:5
    xlabel = (0.2 + 0.12n)
    ylabel = xlabel^n
    plot!(
        x -> x^n; 
        lw=3,
        color=:auto,
        annotation=(xlabel, ylabel, n),
        annotationfontsize=25)
end
plot!(
    ; 
    legend=false, 
    xguide="x", 
    yguide="y", 
    guidefontsize=18, 
    titele=L"x^n \textrm{~labeled~by~}n", 
    titlefontsize=30)
```
