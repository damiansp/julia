```julia
function ginger(x, y, a)
    x2 = 1. - y + a*abs(x)
    #y2 = x
    x2, x
end
```

```julia
x = [20.]
y = [9.]
for i in 1:5000
    x2, y2 = ginger(x[end], y[end], 1.74)
    push!(x, x2)
    push!(y, y2)
end
```

```julia
scatter(x, y, ms=0.5, legend=false)
```
