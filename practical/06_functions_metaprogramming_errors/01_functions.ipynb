```julia
function addthree(a, b, c)
    a + b + c
end
```

```julia
addthree(1, 10, 100)
```

```julia
v3 = (1, 2, 3)
addthree(v3...)
```

```julia
function addthree_coef(a, b, c; beta1=1, beta2=1, beta3=1)
    return beta1*a + beta2*b + beta3*c
end
```

```julia
coefs = (beta1=-1, beta2=2, beta3=3)
addthree_coef(1, 2, 3; coefs...)
```

```julia
coefdict = Dict(:beta1 => -1, :beta2 => 2, :beta3 => -3)
addthree_coef(1, 2, 3; coefdict...)
```

```julia
function addonlythree_with_note(a, b, c, more...)
    if length(more) > 0
        println("Ignoring $(length(more)) superfluous args")
    end
    a + b + c
end
```

```julia
addonlythree_with_note(1, 2, 3, 1, 11, 111)
```

```julia
function examine_string(s; checks...)
    if :palindrome in keys(checks)
        if s == reverse(s)
            println("'$s' is a palindrome")
        end
    end
    if :onlyascii in keys(checks)
        if isascii(s)
            println("'$s' is ASCII only")
        else
            println("'$s' contains non-ASCII chars")
        end
    end
end
```

```julia
examine_string("step on no pets"; kw1=17, palindrome=1, onlyascii=1)
```

```julia
examine_string("step on no pets"; palindrome=1)
```
