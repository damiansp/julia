```julia
bookprice = 24.95
discount = 0.4
shipping1 = 3
shipping_addtl = 0.75
```

```julia
function wholesale(price, disc, n)
    price *= (1 - disc)
    price *= n
    shipping = 3 + 0.75 * (n - 1)
    price + shipping
end
```

```julia
wholesale(bookprice, discount, 60)
```
