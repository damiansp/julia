```julia
function has_no_e(word)
    for letter in word
        if letter == 'e'
            return false
        end
    end
    true
end
```

```julia
function has_no_forbidden_chars(word, forbidden_chars)
    for letter in word
        if letter ∈ forbidden_chars
            return false
        end
    end
    true
end
```

```julia
function usesonly(word, available)
    for letter in word
        if letter ∉ available
            return false
        end
    end
    true
end
```

```julia
function usesall(word, required)
    for letter in required
        if letter ∉ word
            return false
        end
    end
    true
end
```

```julia
function usesall2(word, required)
    usesonly(required, word)
end
```

```julia
function is_abcedarian(word)
    i = firstindex(word)
    prev = word[i]
    j = nextind(word, i)
    for c in word[j:end]
        if c < prev
            return false
        end
    end
    true
end
```

```julia
function is_palindrome(word)
    i = firstindex(word)
    j = lastindex(word)
    while i < j
        if word[i] != word[j]
            return false
        end
        i = nextind(word, i)
        j = prevind(word, j)
    end
    true
end
```
