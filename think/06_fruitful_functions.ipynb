```julia
function area(radius)
    π * radius^2
end
```

```julia
area(7)
```

```julia
function distance(x1, y1, x2, y2)
    dx = x2 - x1
    dy = y2 - y1
    dsq = dx^2 + dy^2
    sqrt(dsq)
end
```

```julia
distance(0, 0, 3, 4)
```

```julia
function hypoteneuse(s1, s2)
    sqrt(s1^2 + s2^2)
end
```

```julia
hypoteneuse(3, 4)
```

```julia
function circle_area(cx, cy, px, py)
    r = distance(cx, cy, px, py)
    area(r)
end
```

```julia
function fact(n)
    if !(n isa Int64)
        error("Factorial only defined over integers.")
    elseif n < 0
        error("Factorial not defined over negatives.")
    elseif n == 0
        return 1
    else
        return n * fact(n - 1)
    end
end
```

```julia
fact("fred")
```

```julia
fact(-6)
```

```julia
fact(7)
```
