```julia
function distance(x1, y1, x2, y2)
    dx = x2 - x1
    dy = y2 - y1
    dsq = dx^2 + dy^2
    sqrt(dsq)
end
```

```julia
distance(0, 0, 3, 4)
```

```julia
function hypoteneuse(s1, s2)
    sqrt(s1^2 + s2^2)
end
```

```julia
hypoteneuse(3, 4)
```
