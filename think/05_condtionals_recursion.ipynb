```julia
function countdown(n)
    if n <= 0
        println("Blastoff!")
    else
        print(n, " ")
        countdown(n - 1)
    end
end
```

```julia
countdown(5)
```

```julia
function printn(s, n)
    if n <= 0
        return 
    end
    println(s)
    printn(s, n - 1)
end
```

```julia
printn("yo", 3)
```

```julia
function checkfermat(a, b, c, n)
    if n > 2 && a^n + b^n == c^n
        println("Fermat was wrong!")
    else
        println("Nope")
    end
end
```

```julia
function is_triangle(a, b, c)
    if a >= b + c
        false
    elseif b >= a + c
        false
    elseif c >= a + b
        false
    else
        true
    end
end
```

```julia
is_triangle(1, 1, 1)
```

```julia
is_triangle(1, 1, 2)
```
