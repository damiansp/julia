```julia
idy((x, y)) = [x, y]
lin1((x, y)) = [2x + 3y, -5x + 4y]
scalex(alpha) = ((x, y),) -> (alpha * x, y)
scaley(alpha) = ((x, y),) -> (x, alpha * y)
rot(theta) = ((x, y),) -> [cos(theta)*x + sin(theta)*y, -sin(theta)*x + cos(theta)*y]
shear(alpha) = ((x, y),) -> [x + alpha*y, y]
genlin(a, b, c, d) = ((x, y),) -> [a*x + b*y; c*x + d*y]
```

```julia
function warp(alpha)
    ((x, y),) -> begin
        r = sqrt(x^2 + y^2)
        theta = alpha * r
        rot(theta)([x, y])
    end
end
```

```julia
rtheta(x) = (norm(x), atan(x[2], x[1]))
xy((r, theta)) = (r * cos(theta), r * sin(theta))
```

```julia
warp2(alpha, x, y) = rot(alpha * sqrt(x^2 + y^2))
warp2(alpha) = ((x, y),) -> warp2(alpha, x, y)([x, y])
```

```julia
warp(1)([5, 6])
```

```julia
warp2(1.)([5., 6.])
```
